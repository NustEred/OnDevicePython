# 데이터 분석

## 1. 데이터 (자료)

- 데이터 : 사실이나 값의 집합. 의미가 없음. 가공되지 않은 원래의 상태값
- 정보 : 데이터를 가공하여 의미 있는 형태로 변환한 결과물

### 데이터의 종류
- 수치 데이터(양적 데이터)
    - 이산형 데이터 : 불량품 수, 판매량
    - 연속형 데이터 : 키, 몸무게, 시간 등
- 범주 데이터 (질적 데이터)
    - 명목형 데이터 : 카테고리, 성별, 지역과 같이 범주나 그룹으로 분류할 수 있는 값
    - 순위형 데이터 : 등급, 선호도, 

## 2. 데이터 분석의 절차

### (1) 문제 정의
- 분석할 문제를 명확히 정의하는 단계
    - 분석의 목적은 무엇인가?
    - 어떤 문제를 해결하려고 하는가?
    - 예상되는 결과는 무엇인가?
    - 어떤 데이터가 필요한가?

### (2) 데이터 수집
- 문제가 정의되면, 이를 해결하기 위해 필요한 데이터를 수집하는 단계
- 데이터는 여러 출처에서 수집될 수 있으며, 데이터 품질과 적절성이 중요

#### 데이터 수집 방법

1차 데이터 수집
- 연구자가 직접 수집한 데이터로 특정 목적을 위해 수집
- 시간과 비용이 많이 소요
- 1. 설문조사 : 대규모 데이터를 비교적 쉽게 수집, 응답률이 낮을 수 있으며 설문지에 따라 데이터 품질이 다름
- 2. 인터뷰 : 상세하고 깊이 있는 데이터를 얻을 수 있음, 시간과 비용이 많이 들며 인터뷰어의 편겨이 개입될 수 있음
- 3. 관찰 : 실제 상황에서의 데이터를 직접 관찰, 특정 상황에 대한 데이터만 수집됨
- 4. 실험 : 원리과 결과의 관계를 명확히 파악, 실험 환경과 실제가 다를 수 있으며 시간과 비용이 많이 소요

2차 데이터 수집
- 이미 수집된 데이터를 활용하는 방법
- 비용과 시간이 절약
- 데이터의 품질과 적합성을 확인하는 것이 중요
- 공공데이터 : 신뢰할 수 있고, 대규모 데이터를 무료로 이동, 데이터의 형식이나 구조가 분석 목적에 맞지 않을 수 있음
- 상업데이터 : 상세하고 특화된 데이터 수집 가능, 비용 소요
- 내부데이터 : 조직에 특화된 데이터, 품질이 일괄되지 않을 수 있고 전처리가 발생
- 온라인데이터 : 실시간 수집, 다양한 형식의 데이터 수집 가능. 수집 방법이 복잡하고, 법적 제한이 있을 수 있음.

## 3. 데이터 전처리

수집한 데이터를 분석에 적합한 형태로 가공, 정제하고 변환하는 과정

### (1) 결측치 처리
- 결측치란 수집한 데이터에서 값이 누락된 경우
- 결측 데이터를 제거하거나 적절한 값으로 대체
- 결측치가 있으면 분석 결과가 왜곡될 수 있음

### (2) 이상치 처리
- 이상치랑 일반적인 데이터와 현저하게 다른 값
- 분석 결과가 왜곡될 수 있음
- 제거하거나 적절한 값으로 대체
- 이상치 탐지 방법 : 시각적 방법 (박스 플롯), 통계적 방법(IQR방법)

### (3) 데이터 변환
- 데이터 분석이나 모델링을 위해 데이터를 다른 형식이나 구조로 변환
- 1. 데이터 분포 조정 : 최소최대값으로 고정(minmax), 표준편차사용(스케일조정)
- 2. 인코딩 : 명목형(원핫인코딩), 순위형(레이블인코딩)

### (4) 데이터 통합
- 여러 소스에서 수집된 데이터를 일관된 형식으로 결합
- 병합 (Merge), 결합(Concate), 변환(Transformation), 중복제거(Duplicate)

## 4. 탐색적 데이터 분석(Exploratory Data Analysis, EDA)
- 전처리된 데이터를 통계적 분석을 통하거나 데이터 시각화를 통하여 데이터를 이해하는 단계
- 데이터 시각화: 히스토그램, 박스플롯, 산점도 등 다양한 그래프를 이용한 분석
- 상관 분석: 변수 간의 상관 관계 파악



## 5. 데이터 모델링
- 수학적 모델을 사용하여 데이터를 분석하고 예측하는 단계
- 동계적 기법과 기계학습 알고리즘을 적용
- 모델링 설치
    - 모델 선택 : 문제에 적합한 모델 선택(예 : 회기 분석, 분류, 클러스터링 등)
    - 모델 학습 : 선택한 모델을 데이터에 맞춰 학습
    - 모델 평가 : 모델의 성능 평가(예: 정확도, 정밀도 재현율 등)

## 6. 모델 평가 및 검증
- 모델의 성능을 평가하고 검증하는 단계
- 교차 검증 : 데이터를 여러 부분으로 나누어 모델을 반복적으로 학습하고 평가
- 성능 지표 : 정확도, 정밀도, F1점수 등 성능 지표 계산
- 오버피팅/언더피팅(과적합) : 모델이 과대적합(overfitting) 또는 과소적합(underfitting)되지 않았는지 확인
