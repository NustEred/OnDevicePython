예제 문제

- 산술연산자 프로그램
- 키보드로 숫자 2개를 입력받아, 산술연산자처리 결과를 출력

예시)
입력 a : 23
입력 b : 15

출력 

a + b = 38
a - b = 8
a * b = 345
a / b = 1.533333
a // b = 1
a % b = 8
a ** b = 6436343

```python
a = int(input("입력 a : "))
b = int(input("입력 b : "))

print(f"a + b = {a+b}")
print(f"a - b = {a-b}")
print(f"a * b = {a*b}")
print(f"a / b = {a/b}")
print(f"a // b = {a//b}")
print(f"a % b = {a%b}")
print(f"a ** b = {a**b}")
```
