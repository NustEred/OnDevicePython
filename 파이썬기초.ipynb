# 파이썬 기초 문법

## 1. 변수와 자료형

파이썬에서 변수는 데이터를 저장하는 공간입니다. 변수는 값을 할당할 때 자동으로 자료형을 결정합니다. 주요 자료형에는 정수(int), 실수 (float), 문자열 (str), 불리언(bool) 등이 있습니다.

`# 변수선언 및 대입 방법`
`변수명 = 값`

````python
# 변수선언 및 대입
val = 5
# 변수값 출력
print(val, type(val), id(val))
````

````python
# 연산 결과 변수에 저장
val = 5 + 7
# 변수값 출력
print(val)
````

````python
# 변수값을 이용한 연산 값 저장
val1 = 12
val2 = 34
val3 = val1 + val2
print(val3)
````

## 2. 자료형
### 숫자형
* 정수(int): 소숫점이 없는 숫자. 1, -23, 0 등
* 실수(float): 소숫점이 있는 숫자. 3.14, -0.5, 12e11, 13E-4 등
* 8진수 : 0o123, 0O456
* 16진수: 0x78, 0x1A, 0X2F 등

````python
# 숫자형 자료형 예제
val1 = 1
val2 = -23
val3 = 3.14
val4 = -0.5
val5 = 12e-6
val6 = 0.123e4
val7 = 0o456
val8 = 0x89A

print(val1, type(val1), id(val1))
print(val2, type(val2), id(val2))
print(val3, type(val3), id(val3))
print(val4, type(val4), id(val4))
print(val5, type(val5), id(val5))
print(val6, type(val6), id(val6))
print(val7, type(val7), id(val7))
print(val8, type(val8), id(val8))
````

### 논리형

* 비교연산이나 논리연산의 결과를 표현하는 자료형
* 조건을 만족하면 참(True), 만족하지 않으면 거짓(False)으로 표현
* 반드시 첫 글자는 대문자로 작성해야 함

````python
a = True
b = False
c = True

print(a, type(a), id(a))
print(b, type(b), id(b))
print(c, type(c), id(c))
````

### 산술연산자

* 숫자를 연산하는데 사용하는 연산 기호
* 더하기(+), 뺴기 (-), 곱하기(*), 나누기 (/)를 연산하는 사칙연산자
* 나눗셈의 몫(//)과 나머지(%)를 구하는 연산자
* 승수(거듭제곱)(**)을 구하는 연산자

````python
a = 7
b = 3
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

### 비교연산자
* 두 수의 크기를 비교하는 연산자
* 크다(>), 크거나 같다(>=), 작다(<), 작거나 같다(<=), 같다 (==), 같지 않다 (!=)
* 연산의 결과를 만족하면 참(True), 만족하지 않으면 거짓(False)으로 결과값을 반환

````python
a = 3
b = 4

print(a > b)
print(a >= b)
print(a < b)
print(a <= b)
print(a == b)
print(a != b)
````

### 논리연산자

* 서로 다른 조건의 논리적 연산에 사용되는 연산자
* 연산자에는 and, or, not이  있음

````python
# and 연산자
print(False and False)
print(False and True)
print(True and False)
print(True and True)
````

````python
# or 연산자
print(False or False)
print(False or True)
print(True or False)
print(True or True)
````

````python
# not 연산자
print(not True)
print(not False)
````

## 3. 제어문
 - 프로그램의 일부가 어떤 조건을 만족하는 경우에만 실행되도록 하거나, 특정 처리를 반복적으로 실행시킬 때 사용
 - 제어문에는 조건문(if문)과 반복문 (for문, while문) 이 있음

### 조건문 if
- 조건문은 특정 조건을 만족할 때만 실행되는 코드 블럭을 정의하는데 사용됩니다. 기본 구조는 다음과 같습니다.

```python
- if문 구조
if 조건식: 
    처리1
    처리2
elif 조건식:
    처리3
    처리4
else :
    처리5
    처리6
```

````python
# 조건 분기가 하나인 경우
n = int(input())
if n % 2 == 0 :
    print("짝수")
else :
    print("홀수")
print("프로그램 종료")
````

### While문 구조
while 조건식 :
    반복처리

````python
# 반복조건을 제어하는 변수
n = 0
sum = 0
while n < 100:
    n = n+1
    sum = sum + n

print(sum)
````

````python
# 1~100 사이의 짝수의 갯수
num = 0
count = 0
while num < 100:
    num = num + 1
    if num % 2 == 0:
        count = count + 1

print(count)
````

- while 문 사용시 주의사항
1. 무한루프에 빠지지 않도록 조건문이 False가 되도록 반복문 내에서 조건을 변경해야 함
2. 반드시 반복을 탈출시킬 수 있는 탈출 조건을 만들어야 함
3. 경우에 따라서 무한루프를 만드는 경우, break 문을 사용해서 반복문을 종료해야 함
4. 반복중 일부분을 거넌뛰고 싶을 때는 continue문을 사용하여 다음 반복으로 넘어가도록 할 수 있음

````python
# 반복문 강제 종료

n = 0 
while True:
    n = n + 1
    print(n)
    if n == 10:
        break

print("프로그램 종료")
````

### 반복문 for
- for문은 시퀀스(리스트, 튜블, 문자열 등)의 각 요소를 순화하면서 반복적으로 실행되는 코드 블럭

``` #for문 구조
for 반복변수 in 시퀀스 : 
    반복처리```

````python
# 문자열 길이만큼 반복

for char in "Life is too short.": 
    print(f"[{char}]")
````

````python
sum = 0
for i in range(1,101):
    sum = sum + i
print(sum)
````

### 제어문 중첩

````python
# for문 중첩
for i in range(2,10):
    print("=== %d단 ===\n" % i)
    for j in range(2,10):
        print("%d * %d = %d" % (i, j, (i*j)))
````

````python
# print() 함수 사용시 자동 줄바꿈 없애는 방법
print("첫번째 출력")
print("2번째 출력", end='')
print("3번째 출력", end=' ')
print("4번째 출력", end='\t')
print("5번째 출력", end='.')
````
