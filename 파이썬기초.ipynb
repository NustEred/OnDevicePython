## 2. 자료형

### While문 구조
while 조건식 :
    반복처리

````python
# 반복조건을 제어하는 변수
n = 0
sum = 0
while n < 100:
    n = n+1
    sum = sum + n

print(sum)
````

````python
# 1~100 사이의 짝수의 갯수
num = 0
count = 0
while num < 100:
    num = num + 1
    if num % 2 == 0:
        count = count + 1

print(count)
````

- while 문 사용시 주의사항
1. 무한루프에 빠지지 않도록 조건문이 False가 되도록 반복문 내에서 조건을 변경해야 함
2. 반드시 반복을 탈출시킬 수 있는 탈출 조건을 만들어야 함
3. 경우에 따라서 무한루프를 만드는 경우, break 문을 사용해서 반복문을 종료해야 함
4. 반복중 일부분을 거넌뛰고 싶을 때는 continue문을 사용하여 다음 반복으로 넘어가도록 할 수 있음

````python
# 반복문 강제 종료

n = 0 
while True:
    n = n + 1
    print(n)
    if n == 10:
        break

print("프로그램 종료")
````

### 중간 문제
- 키보드로 입력 (이름, 국어, 영어, 수학)
- 입력받은 정보로 딕셔너리 생성
- 총점, 평균, 학점을 계산해서 딕셔너리에 추가
- 딕셔너리를 리스트에 추가
- 이름이 "q"가 입력되면 반복문 종료

````python
list3 = []
ranks = {'A':90, 'B':80, 'C':70, 'D':60}

while True:
    student = {'name':'홍길동','kor':99,'eng':88,'mat':77,'tot':0,'avg':0,'grade':'A'}
    student['name'] = input("학생 이름:")
    if student['name'] == 'q':
        break
    student['kor'] = int(input("국어 성적:"))
    student['eng'] = int(input("영어 성적:"))
    student['mat'] = int(input("수학 성적:"))
    student['tot'] = student['kor'] + student['eng'] + student['mat']
    student['avg'] = student['tot'] / 3
    for pending in ranks:
        if student['avg'] >= ranks[pending]:
            break
    student['grade'] = pending
    list3.append(student)

for student in list3:
    for key in student:
        print(f"{key} : {student[key]}")
    print()
````

- 사용자 정의 함수 : 사용자(프로그래머)가 직접 기능을 정의해서 사용하는 것

### 사용자 정의 함수 

- 사용자가 직접 정의하여 호출하는 함수

```Python
# 함수 정의
def 함수명(매개변수):
    처리1
    처리2
    return 반환값
```

````python
# 함수 정의 연습
def add(a, b):
    return a + b

add (2, 3)
````

````python
# 매개변서(파라메타)가 없는 함수
def say():
    return 'Hi'

say()
````

````python
# 변환값(리턴값) 없는 함수
def add(a, b):
    print(f"{a}와 {b}의 합은 {a+b}입니다.")

add(5,6)
````

````python
# 매개변수에 디폴트 값 설정
def cal(n1, n2, op='+'):
    if op == '+':
        return n1 + n2
    elif op == '-':
        return n1-n2
    elif op == '*':
        return n1*n2
    elif op == '/':
        return n1/n2

cal(1,2,'-')
````

````python
def sum(*args):
    result = 0
    for i in args :
        result = result + i
    return result

sum(1,2,3,4,5,6,7,8,9,5)
````

````python
# 반환값이 여러개인 경우
def add_and_mul(a,b) :
    return a+b, a*b

add_result, mul_result = add_and_mul(7,8)
print(add_result,mul_result)
````

````python
# 로컬변수(지역변수)와 글로벌변수(전역변수)
a = 1

def vartest(a):
    a = a + 1

vartest(a)
print(a)
````

````python
# 글로벌(전역)변수 선언 방법
a = 1

def vartest():
    global a
    a = a + 1

vartest()
print(a)
````

````python
# 비밀번호 체크 코드를 함수화하여 변경

alphabet = set("abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ")
number = set("1234567890")
special = set("!@#$%&*+-/")
def test(password):
    paset = set(password)
    if (len(password) >= 8
    and len(password) <= 16
    and len(paset & alphabet) > 0
    and len(paset & number) > 0
    and len(paset & special) > 0
    and len(paset & {' '}) == 0):
       return True
    return False

while True:
    new = input("새로운 비밀번호를 입력해주세요.")
    if test(new):
        break
    print("비밀번호 형식이 올바르지 않습니다.")
print("비밀번호가 설정되었습니다.")
````

### 람다(lambda) 함수
- 람다 함수는 간단한 처리를 한 줄로 정의하여 사용할 수 있는 익명 함수입니다.
- 복잡한 처리는 붉능하며, 주로 간단한 연산이나 함수를 인자로 전다할 때 사용됩니다.

```Python
# 람다 함수 정의 방법
함수형 = lambda 매개변수1, 매개변서2, ... : 매개변수를 이용한 표현식
```

````python
# lambda 함수 예제
add = lambda a, b : a + b

add(3,4)
````

````python
def even_make(start, end):
    even_list = []
    for i in range(start, end+1):
        if i % 2 == 0:
            even_list.append(i)
    return even_list

even_make(1,10)
````

````python
en_make = lambda start, end : [ i for i in range(start, end+1) if i % 2 == 0]
en_make(10,20)
````

## 9. 파일 입출력

### 파일 생성함수 open()

```python
# open() 함수 구조
파일객체명 = open(파일명(패스포함), 파일모드)
#파일닫기
파일객체명.close  
```

* 파일모드

파일모드 | 설명
:-----:|-----
r | 읽기모드, 파일이 존재하지 않으면 오류 발생
w | 쓰기모드, 파일이 존재하지 않으면 새로 생성, 존재하면 기존 내용 덮어쓰기
a | 추가모드, 파일이 존재하지 않으면 새로 생성, 존재하면 기존 내용 뒤에 추가

````python
# 파일 쓰기 (w 모드)
f = open("test.txt", "w", encoding='utf-8')
for i in range(1,11):
    data = f'{i}번째 줄입니다.\n'
    f.write(data)
f.close()
````

````python
# 파일 쓰기 (a 모드)
f = open("test.txt", "a", encoding='utf-8')
for i in range(1,11):
    data = f'{i}번째 줄입니다.\n'
    f.write(data)
f.close()
````

````python
# 파일 읽기
f = open('test.txt', 'r', encoding='utf-8')
line = f.readline()
print(line)
f.close()
````

````python
# 파일의 전체 라인 읽어오기
f = open('test.txt', 'r', encoding='utf-8')

while True : 
    line = f.readline()
    if not line :
        break
    print(line)
f.close()
````

````python
# 파일 읽기 시에 줄바꿈 제거해서 읽어오기

f = open('test.txt', 'r', encoding='utf-8')
lines = f.readlines()
print(lines)
for line in lines:
    line = line.strip()
    print(line)
f.close()
````

````python
# 파일 객체를 사용해서 읽어오기
f = open('test.txt', 'r', encoding='utf-8')
for line in f: 
    line = line.strip()
    print(line)
f.close()
````

### csv 파일

````python
# csv 파일 읽어와서 저장하기

data_list = []

f = open("학생.csv", "r", encoding="utf-8")

for line in f.readlines():
    line = line.strip()
    data_record = line.split(',')
    print(data_record)

f.close()
````

````python
# with 문으로 파일 읽고 쓰기
with open('학생.csv', 'r', encoding = 'utf-8') as f:
    for line in f.readlines():
        line = line.strip()
        data_record = line.split(',')
        data_list.append(data_record)

print(data_list)
````

### 통계 용어
- 분산: 중앙에서 얼마나 퍼저있는 정도
- 표준편차: 분산값이 너무 커서 직관적으로 보기 힘들기 때문에 분산에 루트를 씌워서 크기를 줄인 값
- 평균값: 데이터들의 평균
- 중앙값: 데이터를 작은것부터 큰것으로 나열을 해서, 정 중앙에 위치한 값
- 4분위: 데이터의 범위를 4구역으로 나눈 값(1/4분위(25%), 2/4분위(50%), 3/4분위(75%), 4/4분위(100%))

````python
# 정규분포로 데이터 생성
# np.random.normal (loc = 중앙값, scale = 표준편차, size = 데이터수)
arr2 = np.random.normal(70, 15, 30)
print(arr2)

````

````python
# 표준편차
print(f'표준편차 : {np.std(arr2)}')
# 중앙값
print(f'중앙값 : {np.median(arr2)}')
# 평균값
print(f'평균값 : {np.mean(arr2)}')
# 4분위값
print(f'1분위 : {np.percentile(arr2,25)}')
print(f'2분위 : {np.percentile(arr2,50)}')
print(f'3분위 : {np.percentile(arr2,75)}')
print(f'4분위 : {np.percentile(arr2,100)}')
````
