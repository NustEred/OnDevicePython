2. 성적 계산

- 키보트로부터 이름, 국어, 영어, 수학 점수를 입력받아 총점과 평균을 출력

입력 예)  
이름 : 홍길동  
국어 : 90  
영어 : 80  
수학 : 70  

출력 예)  
이름 : 홍길동  
국어 : 90 점  
영어 : 80 점  
수학 : 70 점  
총점 : 240 점  
평균 : 80.0 점  

```python
name = input("이름: ")
korean = int(input("국어 : "))
english = int(input("영어 : "))
math = int(input("수학 : "))

print("이름 : %s" % name)
print("국어 : %d 점" % korean)
print("영어 : %d 점" % english)
print("수학 : %d 점" % math)
print("총점 : %d 점" % (korean+english+math))
print("평균 : %f 점" % ((korean+english+math)/3))
```

3. 우리 아이 성인 예상키
- 엄마와 아빠의 키를 입력받아 태어난 아이의 예상키를 계산하는 프로그램을 작성하시오.
- 계산식   
    아들 = (엄마키(cm) + 아빠키(cm)) / 2 + 6.5  
    딸 = (엄마키(cm) + 아빠키(cm)) / 2 - 6.5

예)  
엄마 키(cm) : 160  
아빠 키(cm) : 175  

출력 예)  
아들의 키는 173.0 cm로 예상됩니다.  
딸의 키는 160.0 cm로 예상됩니다.

```python
mom=float(input("엄마 키(cm) : "))
dad=float(input("아빠 키(cm) : "))

son = (mom + dad) / 2 + 6.5
dau = (mom + dad) / 2 - 6.5

print("아들의 키는 %f cm으로 예상됩니다." % son)
print("딸의 키는 %f cm으로 예상됩니다." % dau)
```

4. BMI 계산기
- 키와 몸무게를 입력받아 BMI를 계산
- BMI 계산식 : 체중(kg) / (키(m) * 키(m))

입력 예)  
키(cm) : 175  
체중(kg) : 80  

출력 예)  
BMI : 26.122448979591837

```python
hei = float(input("키(cm) : "))
wei = float(input("체중(kg) : "))
bmi = wei * 10000 / hei / hei

print("BMI : %f" % bmi)
```

5. 체지방 계산
- 키와 몸무게를 입력받아 체지방량(kg)을 계산
- 계산식 : 체지방량(kg) = (1.1 * 체중(kg)) - (128 * (체중(kg) / 키(cm)))

입력 예)  
키(cm) : 175
체중(kg) : 90


출력 예)  
당신의 키는 175cm 이고, 몸무게는 90kg으로 체지방량은 33.1kg 입니다.

```python
height = float(input("키(cm) : "))
weight = float(input("체중(kg) : "))

fat = (1.1*weight) - (128* (weight/height))

f"당신의 키는 {height}cm 이고, 몸무게는 {weight}kg으로 체지방량은 {fat}kg 입니다."
```
