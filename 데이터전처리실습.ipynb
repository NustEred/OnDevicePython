## 2. 데이터 수집 (데이터 읽어오기)

```python
test_df = pd.read_csv('practice.csv')
test_df
```

```python
# 데이터 구조 파악
test_df.info()
```

```python
# 결측치 확인
test_df.isnull().sum()
```

```python
# 결측률 측정
(test_df.isnull().mean())
```

```python
# 결측지 삭제(행)
test_df_drop_na = test_df.dropna()
test_df_drop_na
```

```python
# 결측지 삭제 (열(컬럼))
test_df_drop_col = test_df.drop(columns=['Score1'])
test_df_drop_col
```

```python
# 걸측지 대체 (중앙값)
test_df_fillna = test_df.copy()
test_df_fillna['Score1'] = test_df_fillna['Score1'].fillna(
    test_df_fillna['Score1'].median().round()
)

test_df_fillna
```
